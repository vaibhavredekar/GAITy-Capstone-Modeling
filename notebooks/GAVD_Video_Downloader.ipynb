```python
import subprocess
import json

url = "https://www.youtube.com/watch?v=TgkxrrhnvlM"

# Dump JSON metadata
result = subprocess.run(
    ["yt-dlp", "--dump-single-json", url],
    capture_output=True,
    text=True
)

info = json.loads(result.stdout)

if info.get("channel_id") == "UCdYhSujD-axqnb0jjXz8UFQ":
    # Download and trim
    subprocess.run([
        "yt-dlp",
        "--download-sections", "*5.0-30.0",
        "-o", "%(title)s.%(ext)s",
        url
    ])
else:
    print("Skipped, not MissionGait")
```

```python
import subprocess

# Video URL to test
url = "https://www.youtube.com/watch?v=TgkxrrhnvlM"

# Target channel ID
target_channel_id = "UCdYhSujD-axqnb0jjXz8UFQ"

# Step 1: Dump JSON metadata
result = subprocess.run(
    ["yt-dlp", "--dump-single-json", url],
    capture_output=True,
    text=True
)

if result.returncode != 0:
    print(f"Failed to fetch metadata for {url}")
else:
    import json
    info = json.loads(result.stdout)
    
    # Step 2: Check channel_id
    if info.get("channel_id") == target_channel_id:
        print(f"Video matches MissionGait! Downloading...")
        subprocess.run([
            "yt-dlp",
            "-o", "%(title)s.%(ext)s",
            url
        ])
    else:
        print(f"Skipping {url} – channel does not match")
```
