# Move video files from the designated folder according to seq and Video_Speed in a csv filde to another folder

```python
import csv
import os
import shutil

CSV_FILE = "merged_summary_enriched_full - merged_summary_enriched_full.csv"
VIDEO_DIR = "../data/output"
NORMAL_DIR = "output_normal"
SLOWMO_DIR = "output_slowmotion"

# Create output folders if they don't exist
os.makedirs(NORMAL_DIR, exist_ok=True)
os.makedirs(SLOWMO_DIR, exist_ok=True)

seq_map = {}

# Read the semicolon-separated CSV
with open(CSV_FILE, newline="", encoding="utf-8") as f:
    reader = csv.DictReader(f, delimiter=";")

    print("Detected columns:", reader.fieldnames)

    for row in reader:
        seq = row["seq"].strip()
        speed = row["Video_Speed"].strip().upper()
        seq_map[seq] = speed

print(f"Loaded {len(seq_map)} sequences from CSV")

# Process video files
for filename in os.listdir(VIDEO_DIR):
    file_path = os.path.join(VIDEO_DIR, filename)

    if not os.path.isfile(file_path):
        continue

    # Get part before first underscore
    seq_from_filename = filename.split("_")[0]

    if seq_from_filename not in seq_map:
        print(f"Skipping {filename} (not found in CSV)")
        continue

    speed = seq_map[seq_from_filename]

    if speed == "SLOWMOTION":
        target_dir = SLOWMO_DIR
    else:
        target_dir = NORMAL_DIR

    target_path = os.path.join(target_dir, filename)

    print(f"Moving {filename} → {target_dir}/")
    shutil.move(file_path, target_path)

print("Done sorting videos.")
```

# Add the full file name to it instead of comparing only the seq

```python
import csv
import os
import shutil

CSV_FILE = "merged_summary_enriched_full - merged_summary_enriched_full.csv"
VIDEO_DIR = "../data/output"
NORMAL_DIR = "output_normal"
SLOWMO_DIR = "output_slowmotion"

os.makedirs(NORMAL_DIR, exist_ok=True)
os.makedirs(SLOWMO_DIR, exist_ok=True)

lookup = {}

with open(CSV_FILE, newline="", encoding="utf-8") as f:
    reader = csv.DictReader(f, delimiter=";")

    print("Detected columns:", reader.fieldnames)

    for row in reader:
        key = (
            f"{row['seq']}_"
            f"{row['cam_view']}_"
            f"{row['gait_event']}_"
            f"{row['dataset']}_"
            f"{row['gait_pat']}"
        ).strip()

        speed = row["Video_Speed"].strip().upper()

        lookup[key] = speed

print(f"Loaded {len(lookup)} entries from CSV")

for filename in os.listdir(VIDEO_DIR):
    file_path = os.path.join(VIDEO_DIR, filename)

    if not os.path.isfile(file_path):
        continue

    name_no_ext = os.path.splitext(filename)[0]

    # Find matching key at start of filename
    matched = False
    for key, speed in lookup.items():
        if name_no_ext.startswith(key):
            target_dir = SLOWMO_DIR if speed == "SLOWMOTION" else NORMAL_DIR

            print(f"Moving {filename} → {target_dir}/")
            shutil.move(file_path, os.path.join(target_dir, filename))

            matched = True
            break

    if not matched:
        print(f"Skipping {filename} (no CSV match)")

print("Done.")
```

# Sort videos in slowmo folder according to the label slowmotion

```python
import csv
import os
import shutil

CSV_FILE = "merged_summary_enriched_full - merged_summary_enriched_full.csv"
VIDEO_DIR = "/Volumes/NO NAME/output_slowmotion"
SLOWMO_DIR = "output_slowmotion"

# Create output folder
os.makedirs(SLOWMO_DIR, exist_ok=True)

slowmo_seqs = set()

# Read CSV (semicolon separated)
with open(CSV_FILE, newline="", encoding="utf-8") as f:
    reader = csv.DictReader(f, delimiter=";")

    for row in reader:
        if row["Video_Speed"].strip().upper() == "SLOWMOTION":
            slowmo_seqs.add(row["seq"].strip())

print(f"Loaded {len(slowmo_seqs)} slow-motion sequences")

# Process videos
for filename in os.listdir(VIDEO_DIR):
    file_path = os.path.join(VIDEO_DIR, filename)

    if not os.path.isfile(file_path):
        continue

    # Extract seq from filename
    seq_from_filename = filename.split("_")[0]

    if seq_from_filename in slowmo_seqs:
        print(f"Moving {filename} → {SLOWMO_DIR}/")
        shutil.move(file_path, os.path.join(SLOWMO_DIR, filename))

print("Done.")
```
